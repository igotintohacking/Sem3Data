## CATEGORIES OF FUNCTION

```python
# NO PARAMETER NO AND NO RERURN TYPE 
def printline():
    name = input("ENTER NAME")
    print(name)
printline()
```

```python
# no parameter no and with return type
def printline():
    name = input("enter name : ")
    return name 
print(printline())
```

```python
# with parameter and no return type 
def printline(name):
    print(name)
print(printline("python1"))
```

```python
# with parameter and with return type
def printline(name):
    return f"Hello {name}"
print(printline("python1"))
```

## WRITE A PROGRAM TO CREATE A FUNCTION NAMED `SPHEREVOLUME` TO FIND VOLUME OF A SPHERE GIVEN ITS RAIDUS AS PARAMETER
$$ sphere volume = 4/3 * pi * r^3 $$


```python
# $$ SPHEREVOLUME = 4/3 - PI * R^3
import math
def spherevolume():
    r = float(input("enter the value of radius"))
    v = (((4/3)*math.pi)*(
        r**3))
    return v
spherevolume()
    
```

## VARIBLE LENGTH ARGUMENT 

```python
def add(a,b):
    return a+b
```

```python
def add(*n):
    total = 0
    for i in n:
        total+=i
    print(total)
```

```python
add(651+65+16+256+14)
```

```python
add(3,4)
add(7,8,8)
add(6,6,6,7,8)
```
